```python
# signal the neuron receives from the outside world or from other neurons.
# We define a vector as input (like a sensory signal).
# Define a weight vector (like synaptic strengths).
# Compute the neuron’s output using dot product.

import numpy as np

# Input vector (e.g., sensory signal)
x = np.array([0.5, 1.2, -0.7])

# Weight vector (synaptic strengths)
w = np.array([0.8, -0.5, 1.0])

# Bias
b = 0.1

# Neuron output (dot product + bias)
z = np.dot(w, x) + b
# Bias shifts the result slightly. Think of it as baseline neuron activity.
# dot=(0.8∗0.5)+(−0.5∗1.2)+(1.0∗−0.7)

print("Neuron output:", z)

# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Neuron output after activation
activated_output = sigmoid(z)
print("Neuron output (after sigmoid activation):", activated_output)
```
